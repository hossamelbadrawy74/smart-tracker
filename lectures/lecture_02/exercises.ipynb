# **A1**

```python
cart = []
cart.append("Latte")
cart.append("Croissant")
cart.append("Tea")

cart.remove("Croissant")
cart.insert(0, "Water")
cart.pop()
cart.sort()

print(cart, len(cart))

```

```python
assert cart == ["Latte", "Water"], "شوف ترتيب الشغل append/remove/insert/pop/sort"
assert len(cart) == 2
print("A1: ممتاز")
```

# **A2**

```python
numbers = [4, 8, 4, 2, 8, 9, 2, 1, 9, 9]

unique = list(set(numbers))
unique.sort()

print(unique)

```

```python
assert sorted(unique) == [1, 2, 4, 8, 9], "set بيمسح التكرار - وبعدين رتب"
print("A2: ممتاز")
```

# **A3**

```python
grades = {
    "math": 88,
    "english": 76
}

grades["math"] = 92
grades["physics"] = 79
del grades["english"]

best = max(grades.values())
print(grades, best)

```

```python
# كود التحقق الذاتي (موجود أصلًا في النوت بوك):
assert sorted(unique) == [1, 2, 4, 8, 9], "set بيمسح التكرار - وبعدين رتّب"
print("A2: ممتاز")
```
