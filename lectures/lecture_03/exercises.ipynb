```python
grade = 85
level = None

if grade >= 90:
    level = "A"
elif grade >= 80:
    level = "B"
elif grade >= 70:
    level = "C"
elif grade >= 60:
    level = "D"
else:
    level = "F"

print(f"الدرجة {grade} -> {level}")
```

```python
assert level in ("A", "B", "C", "D", "F"), "اﻛﺘﺐ ﺷﺮط if/elif ﻛﺎﻣﻞ اﻷول"
assert not (grade >= 90 and level != "A"), ">=90: A"
assert not (80 <= grade < 90 and level != "B"), "80-89: B"
assert not (70 <= grade < 80 and level != "C"), "70-79: C"
assert not (60 <= grade < 70 and level != "D"), "60-69: D"
assert not (grade < 60 and level != "F"), "<60: F"
print("A1: ﻣﻤﺘﺎز")
```

```python
import random

target = random.randint(1, 50)
tries = 0

guess = int(input("خمن رقم من 1 لـ 50: "))
tries += 1

while guess != target:
    if guess > target:
        print("أقل")
    else:
        print("أكثر")
    guess = int(input("خد تخمين جديد: "))
    tries += 1

print(f"برافو! عدد المحاولات: {tries}")
```

```python
assert tries >= 1, "اﻟﻌﺐ اﻟﻠﻌﺒﺔ اﻷول — tries ﻟﺴﮫ ﺻﻔﺮ"
print("A3: ﻣﻤﺘﺎز")
```

```python
scores = [82, 90, 76, 88, 95]
total = 0
count = 0
biggest = scores[0]

for score in scores:
    total += score
    count += 1
    if score > biggest:
        biggest = score

average = total / count
print(total, round(average, 2), biggest)
```

```python
assert total == 431, "total = ﻣﺠﻤﻮع اﻟﺨﻤﺲ درﺟﺎت"
assert count == 5, "count ﺑﯿﺰود ﻣﺮة ﻓﻲ ﻛﻞ دور"
assert biggest == 95, "biggest ﯾﺘﻐﯿﺮ ﺑﺲ ﻟﻮ ﻟﻘﯿﻨﺎ أﻛﺒﺮ"
assert abs(average - 86.2) < 0.001, "average = total / count"
print("A4: ﻣﻤﺘﺎز")
```
