```python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average(self):
        if not self.grades:
            return 0
        total = 0
        for grade in self.grades:
            total += grade
        return total / len(self.grades)

s = Student("Sara", [85, 90, 76])
print(s.name)
```

```python
# تحقق ذاتي:

assert s.name == "Sara", "الاسم اتخزن على الكائن"
assert abs(s.average() - 83.67) < 0.01, "متوسط سارة"

t = Student("Omar", [70, 80])
assert t.name == "Omar", "كل طالب ليه اسمه"
assert abs(t.average() - 75.0) < 0.01, "متوسط عمر"
print("A1: ممتاز")
```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("رصيد غير كاف")
        else:
            self.balance -= amount

acc = BankAccount("Omar", 100)
acc.deposit(50)
```

```python
# تحقق ذاتي:

assert acc.balance == 150, "الإيداع زود الرصيد"

acc.withdraw(40)
assert acc.balance == 110, "السحب خصم"

acc.withdraw(500)
assert acc.balance == 110, "الحماية من السحب الزايد"

assert BankAccount("Sara").balance == 0, "الرصيد الافتراضي صفر"
print("A2: ممتاز")
```

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"{self.title} صفحة {self.pages}"

b = Book("Python", 350)
print(b)
```

```python
# تحقق ذاتي:

assert str(b) == "Python صفحة 350", "النص بالظبط زي المطلوب"

b2 = Book("Pandas", 500)
assert str(b2) == "Pandas صفحة 500", "كتاب تاني"
print("A3: ممتاز")
```

```python
class Classroom:
    def __init__(self, room):
        self.room = room
        self.students = []

    def add(self, student):
        self.students.append(student)

    def best(self):
        if not self.students:
            return None

        top_student = self.students[0]
        for s in self.students:
            if s.average() > top_student.average():
                top_student = s
        return top_student

room = Classroom("A101")
room.add(Student("Sara", [85, 90, 76]))
```

```python
# تحقق ذاتي:


assert room.room == "A101", "اسم القاعة"
assert len(room.students) == 1, "أول طالب اتضاف"

room.add(Student("Omar", [95, 95]))
assert len(room.students) == 2, "ثاني طالب"
assert room.best().name == "Omar", "الأعلى بمتوسط يفوز"
print("A4: ممتاز")
```

```python
class Counter:
    def __init__(self):
        self.n = 0

    def bump(self):
        self.n += 1
        return self.n

c1 = Counter()
c2 = Counter()
print(c1.bump(), c1.bump(), c2.bump())
```
